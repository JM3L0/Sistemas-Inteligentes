```python
class naiveBayes():

    def __init__(self):
        self.classes = 0
        self.mean = []
        self.std = []
        self.prob_class = []

    def fit(self,X,y):        
        self.classes = np.unique(y)
        for i in self.classes:
            data_class = X[y==i]
            self.mean.append(np.mean(data_class,axis=0))
            self.std.append(np.std(data_class,axis=0))
            self.prob_class.append(len(data_class)/len(X))

    def predict(self,X):
        preds = []
        for x in X:
            pred = []
            for i in self.classes:
                pred.append(self.prob_class[i]*np.prod(self.gauss(self.mean[i],self.std[i],x)))
            pred = np.argmax(pred)
            preds.append(pred)
        return preds
    
    def gauss(self,mean,std,x):

        return 1/(np.sqrt(2*np.pi*std**2))*np.e**(-((x-mean)**2)/(2*std**2))


```
